```python
import nbformat
from nbconvert import HTMLExporter
from pathlib import Path
from google.colab import files

# 1. Ruta del notebook original
notebook_path = Path(
    "/content/drive/MyDrive/Tesis/expansion/notebooks/e2e-model-v1-2.ipynb"
)

# 2. Leer el notebook ejecutado
nb = nbformat.read(notebook_path, as_version=4)

# 3. Conservar solamente las salidas de las celdas de código
new_cells = []

for cell in nb.cells:
    if cell.cell_type == "code" and cell.outputs:
        new_cells.append(
            nbformat.v4.new_code_cell(
                source="",
                outputs=cell.outputs,
                execution_count=None
            )
        )

# 4. Crear un notebook nuevo con esas salidas
nb_limpio = nbformat.v4.new_notebook(cells=new_cells)

# 5. Exportar a HTML
exporter = HTMLExporter()
exporter.exclude_input = True
exporter.exclude_input_prompt = True
exporter.exclude_output_prompt = True

html, resources = exporter.from_notebook_node(nb_limpio)

# 6. Guardar el archivo
salida = Path("/content/resultados.html")
salida.write_text(html, encoding="utf-8")

print(f"Celdas con resultados: {len(new_cells)}")
print(f"Tamaño del HTML: {salida.stat().st_size / (1024**2):.2f} MB")

# 7. Descargar el archivo
files.download(str(salida))
```
