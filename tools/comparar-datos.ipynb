```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')

OLD = "/content/drive/MyDrive/Tesis/fase_octubre/outputs/reportes/granger_completo.csv"
NEW = "/content/drive/MyDrive/Tesis/expansion/outputs/reportes/granger_completo.csv"
k = ["pista", "serie_feat", "serie_target", "direccion"]
old, new = pd.read_csv(OLD), pd.read_csv(NEW)
sig = old[old["bh_rechaza"] == True][k]
m = sig.merge(old, on=k).merge(new, on=k, how="left", suffixes=("_oct", "_nov"))
print(m[k + ["n_oct", "n_nov", "lag_oct", "lag_nov",
             "pval_f_oct", "pval_f_nov", "bh_q_oct", "bh_q_nov"]].to_string())

S_OLD = "/content/drive/MyDrive/Tesis/fase_octubre/data/processed/series_completas.csv"
S_NEW = "/content/drive/MyDrive/Tesis/expansion/data/processed/series_completas.csv"
rd = lambda p: pd.read_csv(p, parse_dates=["fecha"]).set_index("fecha")
o, n = rd(S_OLD)["rwa_tipo_treasurybills"], rd(S_NEW)["rwa_tipo_treasurybills"]
c = o.index.intersection(n.index)
print("fechas en común:", len(c))
print("diferencia relativa máxima:", ((o[c] / n[c]) - 1).abs().max())
print("fechas con diferencia > 1 %:", (((o[c] / n[c]) - 1).abs() > 0.01).sum())
```
